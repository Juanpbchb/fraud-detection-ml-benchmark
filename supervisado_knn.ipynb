## caso 1


```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# 1. Cargar el dataset
df = pd.read_csv('df_CC_NO_ED_Desbalanceado_Sin_Outlier.csv')

# 2. Separar variables predictoras y variable objetivo
X = df.drop('fraud_category_5_classes', axis=1)
y = df['fraud_category_5_classes']

# 3. Separar en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# 4. Probar 3 casos de prueba distintos
casos = [
    {'n_neighbors': 3, 'weights': 'uniform', 'p': 2},
    {'n_neighbors': 5, 'weights': 'distance', 'p': 1},
    {'n_neighbors': 7, 'weights': 'uniform', 'p': 2}
]

for i, params in enumerate(casos, 1):
    print(f"\nCaso de prueba {i}: {params}")
    knn = KNeighborsClassifier(**params)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_test)
    
    acc = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    rec = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)
    
    print(f"Accuracy: {acc:.4f}")
    print(f"Precisión: {prec:.4f}")
    print(f"Recall: {rec:.4f}")
    print(f"F1-score: {f1:.4f}")
    print("Reporte de clasificación:\n", classification_report(y_test, y_pred, zero_division=0))
```

El mejor ajuste es el del Caso de prueba 2 ({'n_neighbors': 5, 'weights': 'distance', 'p': 1}), ya que tiene las mejores métricas globales y mejora el desempeño en algunas clases minoritarias.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score, classification_report,
    roc_curve, auc, precision_recall_curve
)
from sklearn.preprocessing import label_binarize
from sklearn.multiclass import OneVsRestClassifier

# 1. Cargar el dataset
df = pd.read_csv('df_CC_SI_ED_Balanceado_Con_Outliers.csv')

# 2. Separar variables predictoras y variable objetivo
X = df.drop('fraud_category_5_classes', axis=1)
y = df['fraud_category_5_classes']

# 3. Separar en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

# Binarizar las etiquetas para multiclase
classes = y.unique()
y_test_bin = label_binarize(y_test, classes=classes)

casos = [
    {'n_neighbors': 3, 'weights': 'uniform', 'p': 2},
    {'n_neighbors': 5, 'weights': 'distance', 'p': 1},
    {'n_neighbors': 7, 'weights': 'uniform', 'p': 2}
]

for i, params in enumerate(casos, 1):
    print(f"\nCaso de prueba {i}: {params}")
    knn = OneVsRestClassifier(KNeighborsClassifier(**params))
    knn.fit(X_train, label_binarize(y_train, classes=classes))
    y_pred = knn.predict(X_test)
    y_pred_label = [classes[np.argmax(row)] for row in y_pred]
    
    acc = accuracy_score(y_test, y_pred_label)
    prec = precision_score(y_test, y_pred_label, average='weighted', zero_division=0)
    rec = recall_score(y_test, y_pred_label, average='weighted', zero_division=0)
    f1 = f1_score(y_test, y_pred_label, average='weighted', zero_division=0)
    
    print(f"Accuracy: {acc:.4f}")
    print(f"Precisión: {prec:.4f}")
    print(f"Recall: {rec:.4f}")
    print(f"F1-score: {f1:.4f}")
    print("Reporte de clasificación:\n", classification_report(y_test, y_pred_label, zero_division=0))
    
    # ROC y Precision-Recall
    y_score = knn.predict_proba(X_test)
    fpr = dict()
    tpr = dict()
    roc_auc = dict()
    precision = dict()
    recall = dict()
    pr_auc = dict()
    for j in range(len(classes)):
        fpr[j], tpr[j], _ = roc_curve(y_test_bin[:, j], y_score[:, j])
        roc_auc[j] = auc(fpr[j], tpr[j])
        precision[j], recall[j], _ = precision_recall_curve(y_test_bin[:, j], y_score[:, j])
        pr_auc[j] = auc(recall[j], precision[j])
    
    # Plot ROC
    plt.figure(figsize=(10, 4))
    for j in range(len(classes)):
        plt.plot(fpr[j], tpr[j], label=f'Clase {classes[j]} (AUC = {roc_auc[j]:.2f})')
    plt.plot([0, 1], [0, 1], 'k--')
    plt.title(f'Curva ROC - Caso de prueba {i}')
    plt.xlabel('False Positive Rate')
    plt.ylabel('True Positive Rate')
    plt.legend()
    plt.show()
    
    # Plot Precision-Recall
    plt.figure(figsize=(10, 4))
    for j in range(len(classes)):
        plt.plot(recall[j], precision[j], label=f'Clase {classes[j]} (AUC = {pr_auc[j]:.2f})')
    plt.title(f'Curva Precision-Recall - Caso de prueba {i}')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.legend()
    plt.show()
```
